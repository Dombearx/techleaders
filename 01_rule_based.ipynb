# 01 — Rule-Based AI

## What *is* artificial intelligence?

Before machine learning existed, programmers wrote AI **by hand**.  
They observed patterns in data and translated them into `IF / ELSE` rules.

This is called a **rule-based system** (or *expert system*).  
It can work surprisingly well — but someone has to write every single rule.

In this notebook we build two small rule-based classifiers:
1. An **animal classifier** based on physical traits
2. A **spam detector** based on suspicious keywords

## Animal Classifier

A biologist gave us these rules:

| Observation | Classification |
|-------------|----------------|
| Has feathers | bird |
| Has scales **and** lives in water | fish |
| Has fur **and** warm-blooded | mammal |
| None of the above | unknown |

We translate those observations directly into Python `if` statements.

```python
def classify_animal(has_feathers, has_scales, has_fur, lives_in_water, warm_blooded):
    """Classify an animal using hand-written rules."""
    if has_feathers:
        return "bird"
    if has_scales and lives_in_water:
        return "fish"
    if has_fur and warm_blooded:
        return "mammal"
    return "unknown"


animals = [
    # name              feathers  scales   fur    water  warm
    ("Eagle",           True,     False,   False, False, True),
    ("Salmon",          False,    True,    False, True,  False),
    ("Dog",             False,    False,   True,  False, True),
    ("Mystery beast",   False,    False,   False, False, False),
]

print("Rule-Based Animal Classifier")
print("=" * 35)
for name, *features in animals:
    result = classify_animal(*features)
    print(f"  {name:<20} → {result}")
```

## Spam Detector

Same idea applied to email text.  
If a message contains **two or more** suspicious keywords, we call it spam.

This is literally how early spam filters worked — and they caught *a lot* of spam!

```python
SPAM_WORDS = ["free", "winner", "click here", "guarantee", "limited offer"]

def is_spam(email_text: str) -> bool:
    """Simple spam detector using keyword rules."""
    text = email_text.lower()
    matches = [word for word in SPAM_WORDS if word in text]
    return len(matches) >= 2


emails = [
    "Hi, are you free for lunch tomorrow?",
    "FREE winner! Click here for your guaranteed limited offer prize!",
    "Your invoice is attached. Please review.",
    "CLICK HERE for a free limited offer — you are the winner!",
]

print("Spam Classifier")
print("=" * 35)
for email in emails:
    label = "SPAM" if is_spam(email) else "OK  "
    preview = email[:45] + "..." if len(email) > 45 else email
    print(f"  [{label}]  {preview}")
```

## Key Lesson

Rule-based systems are:

- ✅ **Transparent** — you can read every decision the program makes
- ✅ **Fast** to build for simple problems
- ❌ **Brittle** — add a new animal and you need new rules
- ❌ **Hard to scale** — real-world problems have thousands of edge cases

**Machine learning** flips this around: instead of writing rules,  
we show the computer thousands of examples and let it *discover* the rules itself.
